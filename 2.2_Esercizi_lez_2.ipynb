### **1. Indicizzazione booleana**

1. Creare un nuovo vettore con tutti i numeri primi tra 0 e 20. I numeri primi compresi tra 0 e 20 sono: **2, 3, 5, 7, 11, 13, 17, 19**.

2. Utilizza un test logico (ad esempio `>`, `<`, `==`, `!=`) per **selezionare tutte le voci del vettore maggiori di 10**.

3. Utilizza un test logico per selezionare **tutti i numeri primi pari** dalla lista.  


```python

import numpy as np

# Creazione del vettore originale
primi = np.array([2, 3, 5, 7, 11, 13, 17, 19])
print("Vettore originale:", primi)

# Creazione della maschera booleana (opzionale, per capire il meccanismo)
maschera_maggiori_10 = primi > 10
print("Maschera booleana (> 10):", maschera_maggiori_10)

# Applicazione della maschera per filtrare i dati
maggiori_10 = primi[primi > 10]
print("Numeri primi > 10:", maggiori_10)


# Test logico per trovare i numeri divisibili per 2
primi_pari = primi[primi % 2 == 0]
print("Numeri primi pari:", primi_pari)


```

### **3. Seleziona il numero più vicino a 0.5 in un array casuale**
Genera un array 10 x 3 con numeri casuali nell'intervallo [0,1].
Per ogni riga, seleziona il numero più vicino a 0.5. Per farlo usa il fancy indexing.

```python
import numpy as np

# 1. Genera un array 10 x 3 con numeri casuali in [0, 1]
np.random.seed(42) 
passo1 = np.random.rand(10, 3)
print("Array Casuale Originale (10x3):\n", passo1)
print("-" * 50)

# 2. Calcola la distanza assoluta da 0.5
distanze = np.abs(passo1 - 0.5)

# 3. Trova l'indice del valore minimo lungo l'asse delle colonne (axis=1)

indici_colonne = np.argmin(distanze, axis=1)
print("Indici delle colonne con il valore più vicino a 0.5 per riga:\n", indici_colonne)
print("-" * 50)

# 4. Applicazione del Fancy Indexing
# Creiamo un array di indici di riga da 0 a 9
indici_righe = np.arange(10)

# Usiamo una coppia di array di indici (indici_righe, indici_colonne) 
# per estrarre esattamente un elemento per riga
valori_vicini = passo1[indici_righe, indici_colonne]

print("Valori finali più vicini a 0.5 per ciascuna riga:\n", valori_vicini)
```

   _(Questo valore rappresenta approssimativamente la tua frequenza cardiaca a riposo, un comune indicatore di salute.)_

3. **Trova la frequenza cardiaca massima** registrata durante la giornata.  
   _(Questo valore è un'indicazione dell'intensità dell'esercizio fisico.)_

4. **Calcoliamo la percentuale di letture effettuate mentre stavi facendo esercizio fisico (frequenza cardiaca sopra a 120)**:  
   - Crea un nuovo array che contiene `True` quando la frequenza cardiaca è **superiore a 120**, e `False` quando è **inferiore o uguale a 120**.
   - Utilizza una funzione di sintesi per calcolare la percentuale di osservazioni in cui la tua frequenza cardiaca era sopra 120.


```python
import numpy as np

# Creazione dell'array con i dati del Fitbit
frequenza_cardiaca = np.array([68, 65, 77, 110, 160, 161, 162, 161, 160, 161, 162, 163, 164, 163, 162, 100, 90, 97, 72, 60, 70])
print("Dati della frequenza cardiaca:", frequenza_cardiaca)

# Calcolo del minimo
fc_minima = np.min(frequenza_cardiaca)
print(f"Frequenza cardiaca minima (a riposo): {fc_minima} BPM")

# Calcolo del massimo
fc_massima = np.max(frequenza_cardiaca)
print(f"Frequenza cardiaca massima (picco): {fc_massima} BPM")

# - Crea l'array booleano (maschera)
fasi_esercizio = frequenza_cardiaca > 120
print("Array booleano (test > 120):\n", fasi_esercizio)

# - Calcola la percentuale usando la media
percentuale_esercizio = np.mean(fasi_esercizio) * 100
print(f"\nPercentuale di letture in fase di esercizio: {percentuale_esercizio:.2f}%")
```

### **7. Analisi di una matrice**
Data la matrice survey con età, reddito e anni di istruzione

survey_matrix = np.array([
    [25, 40000, 10],
    [32, 52000, 12],
    [40, 63000, 14],
    [29, 47000, 11],
    [35, 58000, 13]
])


1. Seleziona dalla matrice solo gli intervistati con **almeno 12 anni di istruzione** utilizzando un test logico.

2. In un'unica riga di codice, seleziona i **redditi** degli intervistati con **almeno 12 anni di istruzione**.

3. In un'unica riga di codice, calcola il **reddito medio** degli intervistati con **almeno 12 anni di istruzione**.


```python
import numpy as np

# Definizione della matrice originale (Colonne: 0 = Età, 1 = Reddito, 2 = Anni Istruzione)
survey_matrix = np.array([
    [25, 40000, 10],
    [32, 52000, 12],
    [40, 63000, 14],
    [29, 47000, 11],
    [35, 58000, 13]
])


# 1. Selezionare solo gli intervistati con almeno 12 anni di istruzione

# Gli anni di istruzione si trovano nella colonna con indice 2 (l'ultima).
# Creiamo una maschera booleana basata sulla terza colonna e filtriamo le righe.
maschera_istruzione = survey_matrix[:, 2] >= 12
intervistati_filtrati = survey_matrix[maschera_istruzione]

print("1. Righe degli intervistati con istruzione >= 12:\n", intervistati_filtrati)
print("-" * 60)


# 2. Selezionare i REDDITI degli intervistati con almeno 12 anni di istruzione

# Il reddito si trova nella colonna con indice 1. 
# Filtriamo le righe con la maschera e isoliamo la colonna 1.
redditi_filtrati = survey_matrix[survey_matrix[:, 2] >= 12, 1]

print("2. Redditi selezionati (in un'unica riga):", redditi_filtrati)
print("-" * 60)


# 3. Calcolare il REDDITO MEDIO di questi intervistati

# Passiamo l'array estratto al punto precedente direttamente a np.mean()
reddito_medio = np.mean(survey_matrix[survey_matrix[:, 2] >= 12, 1])

print(f"3. Reddito medio (in un'unica riga): {reddito_medio:.2f} $")
```
