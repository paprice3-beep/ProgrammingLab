### **1. Trasformare cicli in list comprehension**

Fai tre esempi di cicli producono una lista e poi riscrivi lo stesso codice usando la list comprehension

```python
quadrati = []
for x in range(1, 6):
    quadrati.append(x**2)

print(quadrati)

#comprehension list

quadrati_comp = [x**2 for x in range(1, 6)]

print(quadrati_comp)  

#esempio2

numeri_pari = []
for x in range(1, 11):
    if x % 2 == 0:
        numeri_pari.append(x)

print(numeri_pari)  

#comprehension list
numeri_pari_comp = [x for x in range(1, 11) if x % 2 == 0]

print(numeri_pari_comp)  

#esempio 3

frutta = ["mela", "banana", "kiwi"]
frutta_maiuscola = []

for f in frutta:
    frutta_maiuscola.append(f.upper())

print(frutta_maiuscola)  

#comprehension list

frutta_maiuscola_comp = [f.upper() for f in frutta]

print(frutta_maiuscola_comp)  
```

### **2. Vettore di Numeri Primi**

1. **Crea un vettore** contenente tutti i numeri primi compresi tra **0 e 10**  
   _(Puoi scriverli direttamente nell'array)._

2. **Conta quanti numeri ci sono nel vettore** utilizzando la funzione `len()`.  **Ottieni lo stesso numero** accedendo all'attributo `.size` del vettore.

3. Quale pensi sia il tipo di dato (`dtype`) del vettore?  Prova a rispondere senza eseguire il codice e Verifica la tua risposta accedendo all'attributo `.dtype` del vettore

4. Scrivi l'array usando una list comprehension che controlla che i numeri siano primi. 

```python
#1
import numpy as np

# Creazione del vettore inserendo i numeri direttamente nell'array
vettore_primi = np.array([2, 3, 5, 7])
print("Vettore:", vettore_primi)

#2
# Conteggio con len()
lunghezza_len = len(vettore_primi)

lunghezza_size = vettore_primi.size

print("Lunghezza con len():", lunghezza_len)
print("Lunghezza con .size:", lunghezza_size)

#3
# Verifica del tipo di dato
tipo_dato = vettore_primi.dtype
print("Il tipo di dato (dtype) è:", tipo_dato)

#4

lista_primi = [x for x in range(2, 11) if all(x % i != 0 for i in range(2, int(x**0.5) + 1))]

# Conversione in array NumPy
vettore_comprehension = np.array(lista_primi)

print("Vettore generato con list comprehension:", vettore_comprehension)


```

### **3.  Operazioni su array**
Crea un array **a**, 1D (senza digitarlo esplicitamente), e genera due nuovi array **b**, contenente una sottostringa a piacere e **c** che è il reverse dell'array **a** (esempio [1,2,3] diventa [3,2,1]) .  Dividi l'array **a** per l'array **c**. Fai la stessa cosa per una lista.

```python
import numpy as np

# 1. Crea un array 1D senza digitarlo esplicitamente (es. numeri da 1 a 5)
a = np.arange(1, 6)  # [1, 2, 3, 4, 5]

# 2. Genera l'array b 
b = a[0:3]           # [1, 2, 3]

# 3. Genera l'array c che è il reverse di a 
c = a[::-1]          # [5, 4, 3, 2, 1]

# 4. Dividi l'array a per l'array c 
risultato_array = a / c

print("Array a:", a)
print("Array b (sotto-array):", b)
print("Array c (reverse):", c)
print("Divisione a / c:", risultato_array)
```
